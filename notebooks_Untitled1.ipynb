```python
import itertools 
import numpy as np

def tsp_path(distance_matrix):
    n = len(distance_matrix)
    cities = list(range(1,n))
    min_cost = float("inf")
    best_path = None

    for perm in itertools.permutations(cities):
        path = [0] + list(perm) + [0]
        cost = sum(distance_matrix[path[i]][path[i+1]] for i in range(n))
        if cost < min_cost:
            min_cost = cost
            best_path = path
        return best_path, min_cost 
```
